```python
from datetime import datetime

def get_financial_quarter(date_str):
    date = datetime.strptime(date_str, "%Y-%m-%d")
    month = date.month
    year = date.year

    # Calculate quarter: April is month 1 of financial year
    quarter = ((month - 4) % 12) // 3 + 1
    fin_year = year + 1 if month >= 4 else year

    return f"Q{quarter}FY{str(fin_year)[-2:]}"
```

```python
def get_data_preprocessed(sheet_name='Price to Sales Values',dates_parsing=True):
    df_temp = xls.parse(sheet_name=sheet_name,skiprows=5)
    df_temp.drop(columns=['Unnamed: 0', 'ISIN/BB Code'],inplace=True)
    cols = list(df_temp.columns)
    s = pd.Series(cols)
    s = s.apply(lambda x: np.nan if 'Unnamed' in x else x).ffill()
    if len(df_temp.columns)==len(s):
        print("Validation1")
    df_temp.columns = s
    df_temp2 = pd.DataFrame(columns=['Ticker','Date','Value'])


    for i in set(df_temp.columns):
        print(i)
        df_temp1 = df_temp[i][1:]
        if df_temp1.shape[1]>2:
            df_temp1 = df_temp1.iloc[:,0:2]
        df_temp1.columns = ['Date','Value']
        # df_temp1['Date'] = pd.to_datetime(df_temp1['Date'])
        df_temp1 = df_temp1[~df_temp1['Date'].isna()]
        df_temp1['Ticker'] = i
        df_temp1 = df_temp1.pivot_table(index=['Ticker'],values=['Value'],columns=['Date'])
        df_temp1.columns = df_temp1.columns.get_level_values(1)
        df_temp1.reset_index(inplace=True)
        df_temp2 = pd.concat([df_temp2,df_temp1],axis=0)    

    df_temp2.drop(columns=['Date','Value'],inplace=True)
    df_temp3 = df_temp2.set_index(keys=['Ticker']).sort_index(axis=1)
    df_temp3.reset_index(drop=False,inplace=True)
    if dates_parsing:
        date_cols = list(df_temp2.columns)
        date_cols.remove("Ticker")
        date_cols1 = pd.Series(date_cols).astype(str).apply(get_financial_quarter)
        dict(zip(date_cols,date_cols1))
        df_temp3.rename(columns=dict(zip(date_cols,date_cols1)),inplace=True)
    return df_temp3
```
