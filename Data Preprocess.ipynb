# BETA

```python
df_price_momentum_2= pd.read_excel(r"C:\Users\43737\Desktop\Data for claude\Monthly Price Data Dec'24_Aug'25.xlsx", sheet_name = "Sheet1", index_col= None)
```

```python
df_price_momentum_2.columns = [
    pd.to_datetime(c) if c != 'Ticker' else c
    for c in df_price_momentum_2.columns
]
```

```python
df_prices = df_price_momentum_2.set_index('Ticker')

df_monthly_rets = df_prices.pct_change(axis=1)
```

```python
df_price_momentum_2.head(2)
```

```python
q1fy26_cols = df_monthly_rets.columns[
    (df_monthly_rets.columns >= pd.Timestamp('2025-05-30')) &
    (df_monthly_rets.columns <= pd.Timestamp('2025-08-29'))
]
df_q1fy26_rets = df_monthly_rets[q1fy26_cols]
```

```python
df_q1fy26_rets_t = df_q1fy26_rets.T
```

```python
df_cov = df_q1fy26_rets_t.cov()
```

```python
df_cov_with_index = df_cov['nsemcap'].reset_index()
df_cov_with_index.columns = ['Ticker', 'Covariance_Q1FY26']
```

```python
index_var = df_q1fy26_rets_t['nsemcap'].var()
```

```python
df_beta = df_cov_with_index.copy()
df_beta['Var_Q1FY26'] = index_var
df_beta['Beta_Q1FY26'] = df_beta['Covariance_Q1FY26'] / df_beta['Var_Q1FY26']
```

```python
df_data = df_data.merge(
    df_beta[["Ticker", "Beta_Q1FY26"]],
    on="Ticker",
    how="left"
)
```
